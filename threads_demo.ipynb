##**Script that runs 3 threads simultaneously, each printing a counter 0–9**

```python
%%writefile threading_demo.py
import threading
import time

def print_counter(thread_name):
    """
    Target function for our threads. Loops from 0 to 9
    and prints the counter along with the thread's name.
    """
    for i in range(10):
        print(f"[{thread_name}] Counter: {i}")
        # A tiny sleep forces Python to pause slightly, making the
        # overlapping behavior easier to see in the terminal.
        time.sleep(0.01)

# --- Main Program Control Flow ---
if __name__ == "__main__":
    print("Main program started. Creating 3 concurrent threads...")

    # 1. Initialize 3 separate threads, passing a unique name to each
    thread_1 = threading.Thread(target=print_counter, args=("Thread-A",))
    thread_2 = threading.Thread(target=print_counter, args=("Thread-B",))
    thread_3 = threading.Thread(target=print_counter, args=("Thread-C",))

    # 2. Fire up all 3 threads simultaneously
    thread_1.start()
    thread_2.start()
    thread_3.start()

    # 3. Force the main program to wait until all 3 threads finish their loops
    thread_1.join()
    thread_2.join()
    thread_3.join()

    print("All threads completed. Main program exiting cleanly.")
```
