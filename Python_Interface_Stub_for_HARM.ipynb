```python
from abc import ABC, abstractmethod
from typing import Any, Dict, NamedTuple

class HARMResult(NamedTuple):
    """
    A structured, immutably bound container representing the frozen evaluation
    status of a parsed VLM kinematic output response.
    """
    is_valid: bool
    error_type: str | None
    validated_coordinates: Dict[str, float] | None


class FallbackCommand(NamedTuple):
    """
    A structured data container specifying the safe recovery trajectory vector
    injected to protect the physical hardware during an active system fault.
    """
    action: str
    coordinates: Dict[str, float]
    deceleration_factor: float


class AbstractHARM(ABC):
    @abstractmethod
    def validate(self, vlm_response: str) -> HARMResult:
        """
        Interceptors raw textual string outputs from the VLM Cognitive Engine
        and runs a deterministic 3-stage validation pipeline.

        Parameters:
        -----------
        vlm_response : str
            The unparsed raw text block or JSON string generated by the VLM.

        Returns:
        --------
        HARMResult
            A named tuple indicating success status, the exact failure code
            if invalid, and the extracted Cartesian float positions if safe.

        Raises:
        -------
        TypeError
            If the incoming vlm_response argument is not a primitive string type.
        """
        pass

    @abstractmethod
    def recover(self, error_type: str) -> FallbackCommand:
        """
        Executes an automated algorithmic error-correction or trajectory clipping
        routine based on the intercepted system anomaly type.

        Parameters:
        -----------
        error_type : str
            The exact classification string identifying the failure mode
            (e.g., 'JSONFormattingError' or 'WorkspaceBoundViolation').

        Returns:
        --------
        FallbackCommand
            A structured object detailing a safe, hardware-compliant fallback movement
            such as a hold-in-place state, a re-prompting task, or spatial vector clipping.

        Raises:
        -------
        KeyError
            If an unknown or unregistered error_type token is supplied, preventing
            the module from determining a valid recovery path.
        """
        pass
```
