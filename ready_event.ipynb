```python
import threading
import time

# 1. Initialize the ready event flag (Starts as False/Red Light)
ready_event = threading.Event()

def dependent_worker():
    """
    Thread A: Represents a process that absolutely requires
    another system component to be initialized first.
    """
    print(f"[{threading.current_thread().name}] Thread spawned. Waiting for initialization signal...")

    # 2. Block here efficiently. Uses 0% CPU while waiting for the flag to turn True.
    ready_event.wait()

    # This line can only execute AFTER ready_event.set() is called
    print(f"[{threading.current_thread().name}] Signal received! Starting processing tasks safely.")

def initializer_worker():
    """
    Thread B: Prepares resources, boots hardware, or calibrates sensors.
    """
    print(f"[{threading.current_thread().name}] Hardware bootup sequence initiated...")

    # Simulate calibration/setup latency
    time.sleep(2.5)

    print(f"[{threading.current_thread().name}] Hardware calibration complete. System is READY!")

    # 3. Raise the flag to turn the light green and wake up any waiting threads
    ready_event.set()

# --- Execution ---
# Create Thread A and Thread B
thread_a = threading.Thread(target=dependent_worker, name="Motor_Controller_A")
thread_b = threading.Thread(target=initializer_worker, name="Sensor_Calibrator_B")

# Start them up
thread_a.start()
time.sleep(0.5) # Small delay just to show Thread A waiting first
thread_b.start()

# Wait for both to finish completely
thread_a.join()
thread_b.join()

print("[Main] All synchronization tasks finalized.")
```
