```python
import threading
import time

# 1. Initialize the stop event flag (Starts as False/Down)
stop_event = threading.Event()

def continuous_worker():
    """
    A background thread simulation that runs continuously
    until it is signaled to stop.
    """
    print(f"[{threading.current_thread().name}] Background worker thread started.")

    # Keep looping as long as the stop_event flag is NOT set
    while not stop_event.is_set():
        print(f"[{threading.current_thread().name}] Monitoring system status... (Working)")
        time.sleep(0.5)  # Simulate a brief work interval

    # Once stop_event.is_set() becomes True, the loop breaks and reaches here:
    print(f"[{threading.current_thread().name}] Stop signal detected! Cleaning up hardware resources...")
    time.sleep(0.2)
    print(f"[{threading.current_thread().name}] Thread safely shut down.")

# 2. Create and start the worker thread
worker_thread = threading.Thread(target=continuous_worker, name="Sensor_Monitor")
worker_thread.start()

# 3. Let the main program run alongside it for a few seconds
print("[Main] Letting the worker run for 2 seconds...")
time.sleep(2.0)

# 4. Trigger the stop pattern!
print("\n[Main] Target duration reached. Signaling the worker thread to stop now!")
stop_event.set()  # Turns the flag to True

# 5. Wait for the worker thread to finish its cleanup completely
worker_thread.join()
print("[Main] Execution finalized. Everything is safe.")
```
